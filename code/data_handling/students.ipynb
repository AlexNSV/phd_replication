## MAIN STUDENTS

```python
NATIONAL_DATA_PATH = "../../data/raw/students/OPRI/OPRI_DATA_NATIONAL.csv"
METADATA_PATH = "../../data/raw/students/OPRI/OPRI_METADATA.csv"
LABEL_PATH = "../../data/raw/students/OPRI/OPRI_LABEL.csv"
```

```python
def load_students_data():
    national_data = pd.read_csv(NATIONAL_DATA_PATH, dtype=str).astype(str)
    metadata = pd.read_csv(METADATA_PATH, dtype=str).astype(str)
    labels = pd.read_csv(LABEL_PATH, dtype=str).astype(str)
    national_data['INDICATOR_LABEL_EN'] = national_data.merge(labels, on = 'INDICATOR_ID')['INDICATOR_LABEL_EN']
    # to keep Total inbound internationally mobile students, both sexes (number)
    national_data['INDICATOR_LABEL_EN'] = national_data['INDICATOR_LABEL_EN'].apply(lambda x: x.lower())
    national_data = national_data[national_data['INDICATOR_LABEL_EN'].str.contains('inbound internationally mobile students', na=False)]
    national_data['country_of_origin'] = national_data['INDICATOR_LABEL_EN'].str.extract(r': students from (.+?), both sexes', expand=False)
    
    #national_data = national_data.dropna(subset=['country_of_origin'])
    national_data = national_data.fillna('')
    national_data['country_of_origin'] = national_data['country_of_origin'].apply(lambda x: x.replace('the ', ''))
    return national_data, metadata, labels
```

```python
national_data, metadata, labels = load_students_data()
```

```python
national_data.to_csv('../../data/basic_preprocessed/students.csv')
```
