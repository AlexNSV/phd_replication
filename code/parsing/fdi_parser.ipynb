```python
DEFAULT_TIMEOUT=5
proxy_fails = {}
def try_country(supplier_code=None, reciever_code=None, supplier_name="Unknown", reciever_name="Unknown", direction='inward',proxy=None, bycountries=True, source='investmentmap'):
    global proxy_fails
    if source=='investmentmap':
        if bycountries:
            url = f"https://www.investmentmap.org/api/fdi/time-series/country/excel?countryCode={supplier_code}&partnerCountryCode={reciever_code}&economicActivityCode=TT&type=stock&direction={direction}&startYear=2000&endYear=2023&locale=en"
        else:
            url = f"https://www.investmentmap.org/api/fdi/time-series/economic-activity/excel?countryCode={supplier_code}&partnerCountryCode={reciever_code}&economicActivityCode=TT&type=stock&direction={direction}&startYear=2000&endYear=2023&locale=en"
    elif source=='unctad':
        url = f"https://unctad.org/system/files/non-official-document/webdiaeia2014d3_{supplier_code}.xls"
    else:
        raise ValueError("UNKNOWN SOURCE")
    try:
        print(f'Trying supplier_code={supplier_code}, reciever_code={reciever_code}, proxy={proxy is not None}')
        print(url)
        response=requests.get(url, headers=headers, proxies=proxy, timeout=DEFAULT_TIMEOUT)
        print(f'supplier_code={supplier_code}, reciever_code={reciever_code} response status={response.status_code}, proxy={proxy is not None}')
        if response.status_code != 200: raise Exception(f"status code {status_code}")
        if source == 'investmentmap':
            with open(f'{IMAP_DATA}/fdi{direction[0:2]}_{supplier_code}_{supplier_name}_{reciever_code}_{reciever_name}.xlsx', 'wb') as f:
                f.write(response.content)
        elif source == 'unctad':
            with open(f'{UNCTAD_DATA}/fdi_{supplier_code}.xls', 'wb') as f:
                f.write(response.content)
    except Exception as E:
        print(E.__str__())
        if proxy is not None:  # retrying once without proxy
            if str(proxy) in proxy_fails.keys():
                proxy_fails[str(proxy)] += 1
            else:
                proxy_fails[str(proxy)] = 1
            #print(f'Retrying with no proxy for supplier_code={supplier_code}, reciever_code={reciever_code}')
            try_country(supplier_code=supplier_code, reciever_code=reciever_code, supplier_name=supplier_name, reciever_name=reciever_name, proxy=None, bycountries=bycountries, source=source)
    sleep(1 + 1/random()/10)
    if proxy is None: sleep(5)
```
